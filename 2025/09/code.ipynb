```python
type Coordinate = tuple[int, int]

def parse_input(lines: Iterable[str]) -> list[Coordinate]:
    return list(tuple(map(int, line.split(','))) for line in lines)

parsed_input = parse_input(part1_example_input.splitlines())
# parsed_input = parse_input(inputdata.splitlines())
```

```python
def rectangle_area(corner_a: Coordinate, corner_b: Coordinate) -> int:
    return (abs(corner_a[0] - corner_b[0])+1)*(abs(corner_a[1] - corner_b[1]) + 1)
```

```python
from itertools import combinations


type CoordinatePair = tuple[Coordinate, Coordinate]

def areas_sorted_desc(red_tiles: Iterable[Coordinate]) -> list[CoordinatePair]:
    return sorted(combinations(red_tiles, 2),
                  key=lambda corners: rectangle_area(*corners),
                  reverse=True)
```

```python
print(rectangle_area(*(areas_sorted_desc(parsed_input)[0])))
```

```python
def compress_coordinates(coords: list[Coordinate]) -> dict[Coordinate, Coordinate]:
    def compress_axis(coords: list[int]) -> dict[int, int]:
        return {c: i*2 for i, c in enumerate(sorted(set(coords)))}
    
    x_compressed = compress_axis([x for x, _ in coords])
    y_compressed = compress_axis([y for _, y in coords])
    return {(x, y): (x_compressed[x], y_compressed[y]) for x, y in coords}
```

```python
compressed_dict = compress_coordinates(parsed_input)
compressed_coords = list(compressed_dict.values())
```

```python
max_col = max(x for x, _ in compressed_coords)
max_row = max(y for _, y in compressed_coords)
```

```python
compressed_map = [['.']*(max_col + 1) for _ in range(max_row + 1)]
```

```python
for red_col, red_row in compressed_coords:
    compressed_map[red_row][red_col] = 'x'
```

```python
def green_fill(red_a: Coordinate, red_b: Coordinate, grid: list[list[str]]) -> None:
    assert((red_a[0] == red_b[0]) or (red_a[1] == red_b[1]))
    if red_a[0] == red_b[0]:
        col = red_a[0]
        for row in range(min(red_a[1], red_b[1]) + 1, max(red_a[1], red_b[1])):
            grid[row][col] = 'o'
    else:
        row = red_a[1]
        for col in range(min(red_a[0], red_b[0]) + 1, max(red_a[0], red_b[0])):
            grid[row][col] = 'o'
```

```python
from itertools import chain, pairwise


for red_a, red_b in chain(pairwise(compressed_coords), [(compressed_coords[-1], compressed_coords[0])]):
    green_fill(red_a, red_b, compressed_map)
```

```python
def pad_grid(grid: list[list[str]]) -> None:
    height = len(grid)
    width = len(grid[0])
    grid.insert(0, ['.']*width)
    grid.append(['.']*width)
    for line in grid:
        line.insert(0, '.')
        line.append('.')
```

```python
pad_grid(compressed_map)
```

```python
print('\n'.join(''.join(line) for line in compressed_map))
```
