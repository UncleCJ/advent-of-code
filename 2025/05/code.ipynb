```python
def parse_input(lines):
    end_of_ranges = False
    parsed_ranges = []
    parsed_items = []
    for line in lines:
        if not end_of_ranges:
            if line == '':
                end_of_ranges = True
                continue
            lower, upper = map(int, line.split('-'))
            parsed_ranges.append((lower, upper))
        else:
            parsed_items.append(int(line))
    return (parsed_ranges, parsed_items)

# parsed_ranges, parsed_items = parse_input(part1_example_input.splitlines())
parsed_ranges, parsed_items = parse_input(inputdata.splitlines())
```

```python
from typing import List, Tuple


def merge_ranges(ranges: List[Tuple[int, int]]) -> List[Tuple[int, int]]:
    merged_ranges = []
    pivot_range_lower, pivot_range_upper = ranges.pop(0)

    while len(ranges) > 0:
        pop_index = 0
        for i, (range_next_lower, range_next_upper) in enumerate(ranges):
            if range_next_lower <= pivot_range_upper + 1:
                pivot_range_upper = range_next_upper
                pop_index = i + 1
            else:
                break
        merged_ranges.append((pivot_range_lower, pivot_range_upper))
        print(f'{merged_ranges=}')
        ranges = ranges[pop_index:]
        print(f'{ranges=}')
        if len(ranges) > 0:
            pivot_range_lower, pivot_range_upper = ranges.pop(0)
    if len(ranges) == 1:
        merged_ranges.append(ranges[0])
    return merged_ranges
        
#print(parsed_ranges[:20])
merged_ranges = merge_ranges(parsed_ranges)   
print(merged_ranges)
        
                
```
