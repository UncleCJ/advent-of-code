```python
def parse_input(lines: Iterable[str]) -> tuple[list[tuple[int, int]], list[int]]:
    end_of_ranges = False
    parsed_ranges = set()
    parsed_items = []
    for line in lines:
        if not end_of_ranges:
            if line == '':
                end_of_ranges = True
                continue
            lower, upper = map(int, line.split('-'))
            parsed_ranges.add((lower, upper))
        else:
            parsed_items.append(int(line))
    return (sorted(parsed_ranges), parsed_items)

parsed_ranges, parsed_items = parse_input(part1_example_input.splitlines())
# parsed_ranges, parsed_items = parse_input(inputdata.splitlines())
```

```python
def merge_ranges(ranges_input: list[tuple[int, int]]) -> list[tuple[int, int]]:
    ranges = ranges_input.copy()
    merged_ranges = []

    while len(ranges) > 0:
        pop_index = 0
        pivot_range_lower, pivot_range_upper = ranges.pop(pop_index)
        for i, (range_next_lower, range_next_upper) in enumerate(ranges):
            if range_next_lower <= pivot_range_upper + 1:
                if range_next_upper >= pivot_range_upper:
                    print(f'merging range\n {pivot_range_lower}-{pivot_range_upper} with\n {range_next_lower}-{range_next_upper}')
                    pivot_range_upper = range_next_upper
                pop_index = i + 1
            else:
                break
        merged_ranges.append((pivot_range_lower, pivot_range_upper))
        ranges = ranges[pop_index:]
    return merged_ranges
        
merged_ranges = merge_ranges(parsed_ranges)   
print(merged_ranges)
```

```python
print(sum(upper+1-lower for lower,upper in merged_ranges))
```
