```python
from typing import Iterable, Tuple

def parse_input(lines: Iterable[str]) -> Tuple[Tuple[Tuple[bool, ...], Tuple[int, ...], Tuple[int, ...]], ...]:
    results = []
    for line in lines:
        tokens = list(line.split(' '))
        indicators = tuple(i == '#' for i in tokens[0][1:-1])
        wirings = tuple(tuple(map(int, token[1:-1].split(','))) for token in tokens[1:-1])
        joltages = tuple(map(int, tokens[-1][1:-1].split(',')))
        # print(f'{tokens=} {indicators=} {wirings=} {joltages=}')
        results.append((indicators, wirings, joltages))
    return tuple(results)

parsed_input = parse_input(part1_example_input.splitlines())
# parsed_input = parse_input(inputdata.splitlines())
```
