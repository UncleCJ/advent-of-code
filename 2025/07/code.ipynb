```python
from typing import Iterable


def parse_input(lines: Iterable[str]) -> tuple[tuple[int, int], set[tuple[int, int]]]:
    splitters_by_row = dict()
    for row, line in enumerate(lines):
        if row == 0:
            entry_column = line.find('S')
            continue
        if row % 2 != 0:
            continue
        splitters_by_row[row] = tuple(col for col, c in enumerate(line) if c == '^')
    return (entry_column, splitters_by_row)

entry_column, splitters_by_row = parse_input(part1_example_input.splitlines())
# entry_column, splitters_by_row = parse_input(inputdata.splitlines())
```

```python
def run_beams(entry_column: int, splitters_by_row: dict[int,tuple[int, ...]]) -> int:
    split_count = 0
    current_beams = {entry_column}
    for row in range(2, max(splitters_by_row.keys()) + 1, 2):  # Let's not assume dictionaries remain sorted, for prudence
        assert(row in splitters_by_row.keys())
        for col in splitters_by_row[row]:
            if col in current_beams:
                split_count += 1
                current_beams.remove(col)
                current_beams.add(col - 1)
                current_beams.add(col + 1)
        print(f'{row=}, {splitters_by_row[row]=}, {current_beams=}') 
    return split_count
```

```python
print(run_beams(entry_column, splitters_by_row))
```

```python
from collections import defaultdict


def run_tachyons(entry_column: int, splitters_by_row: dict[int, tuple[int, ...]]) -> int:
    current_tachyons = defaultdict(int, {entry_column: 1})
    for row in range(2, max(splitters_by_row.keys()) + 1, 2):
        assert(row in splitters_by_row.keys())
        for col in splitters_by_row[row]:
            if col in current_tachyons.keys():
                tachyon_count = current_tachyons[col]
                current_tachyons[col] = 0
                current_tachyons[col - 1] += tachyon_count
                current_tachyons[col + 1] += tachyon_count
        # print(f'{row=}, {splitters_by_row[row]=}, {current_tachyons=}') 
    return sum(current_tachyons.values())
```

```python
print(run_tachyons(entry_column, splitters_by_row))
```
