```python
 

def parse_input(lines: Iterable[str]):
    return tuple(tuple(int(d) for d in line) for line in lines)
from typing import Tuple

def max_joltage(bank: Tuple[int, ...]) -> int:
    tens_digit = max(bank[:-1])
    tens_digit_i = bank.index(tens_digit)
    ones_digit = max(bank[tens_digit_i+1:])
    return tens_digit*10+ones_digit

result = 0
for bank in parsed_input:
    result += max_joltage(bank)
print(result)

from typing import Tuple

def max_joltage_n(n: int, bank: Tuple[int, ...]) -> int:
    results = []
    current_index = -1
    print(f'{bank=}')
    for i, j in zip(range(n),range(n-1,-1,-1)):
        if len(results) > 0:
            print(f'{results[-1]=}, {current_index=}, {i=}, {j=}')
        if j != 0:
            results.append(max(bank[current_index+1:-j]))
        else:
            results.append(max(bank[current_index+1:]))
        current_index = bank.index(results[-1], current_index+1)
    print(f'{bank=}: {results=}')
    total = sum(results[i]*10**j for i,j in zip(range(n), range(n-1, 0, -1)))
    print(total)
    return total
print(sum(max_joltage_n(12, bank) for bank in parsed_input))
 
 
```
