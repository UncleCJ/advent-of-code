```python
from typing import Tuple


def adjust_bounds(lower: str, upper: str) -> Tuple[str, str]:
    if len(lower) % 2:
        lower = str(10**(len(lower)))
    if len(upper) % 2:
        upper = str(10**(len(upper)-1)-1)
    return (lower, upper)

adjusted_input = [adjust_bounds(lower, upper) for lower, upper in parsed_input]
```

```python
validated_input = [(lower, upper) for lower, upper in adjusted_input if int(lower) <= int(upper)]
```

```python
def bisect_number(number: str) -> Tuple[str, str]:
    assert(len(number) % 2 != 1)
    half_len = len(number) // 2
    big = number[:half_len]
    small = number[half_len:]
    return (big, small)
```

```python
from typing import Iterable, List


def find_invalid(lower: str, upper: str) -> Iterable[str]:
    assert(len(lower) == len(upper))
    # print(f'{lower=} - {upper=}')
    lower_int = int(lower)
    upper_int = int(upper)
    lower_big, lower_small = bisect_number(lower)
    upper_big, upper_small = bisect_number(upper)
    lower_big_int = int(lower_big)
    lower_small_int = int(lower_small)
    upper_big_int = int(upper_big)
    upper_small_int = int(upper_small)
    if (lower_big_int >= lower_small_int):
        range_lower = int(lower_big)
    else:
        range_lower = int(lower_big)+1
    
    if (upper_big_int <= upper_small_int):
        range_upper = int(upper_big)
    else:
        range_upper = int(upper_big)-1
    
    for i in range(range_lower, range_upper+1):
        # print(f'{str(i)*2}')
        yield str(i)*2

    return

def flatten(xxs: Iterable[Iterable[str]]) -> List[str]:
    return [x for xs in xxs for x in xs]

candidate_invalids = flatten(find_invalid(lower, upper) for lower, upper in validated_input)
# for invalid in candidate_invalids:
#     print(invalid)
```

```python
from typing import Iterator


def determine_parts_n(n: int, lower: str, upper: str) -> Iterator[Tuple[int, str, str]]:
    for len_candidate in range(len(lower), len(upper)+1):
        if (len_candidate % n == 0):
            if (len_candidate > len(lower)):
                lower = str(10**(len_candidate-1))
            if (len_candidate < len(upper)):
                upper = str(10**len_candidate - 1)

            yield (n, lower, upper)
    
print(*determine_parts_n(2, '88', '110'))

def determine_parts(lower: str, upper: str) -> Iterator[Tuple[int, str, str]]:
    for n in range(2, len(upper) + 1):
        for item in determine_parts_n(n, lower, upper):
            yield item

print(*determine_parts('88', '110'))
print(*determine_parts('824824821','824824827'))
```

```python
def split_string(n: int, s: str) -> Tuple[str, ...]:
    assert(len(s) % n == 0)
    part_len = len(s) // n
    return tuple(s[i*part_len:(i+1)*part_len] for i in range(n))

print(split_string(3, '123456789'))
```

```python
from typing import Set


def find_repeats_n(n: int, lower: str, upper: str) -> Iterator[int]:
    assert(len(lower) == len(upper))
    # print(f'{n=}: {lower=} - {upper=}')
    lower_parts_i = tuple(map(int, split_string(n, lower)))
    upper_parts_i = tuple(map(int, split_string(n, upper)))
    for i in range(1, n):
        if (lower_parts_i[0] < lower_parts_i[i]):
            part_lower = lower_parts_i[0] + 1
            break
    else:
        part_lower = lower_parts_i[0]
    
    for i in range(1, n):
        if (upper_parts_i[0] > upper_parts_i[i]):
            part_upper = upper_parts_i[0] - 1
            break
    else:
        part_upper = upper_parts_i[0]
    
    for i in range(part_lower, part_upper + 1):
        yield int(str(i)*n)

    return

print(*find_repeats_n(2,'1000','1012'))
print(*find_repeats_n(3,'998','999'))

def find_repeats(lower: str, upper: str) -> Set[int]:
    result = set()
    for n, lower, upper in determine_groups(lower, upper):
        for repeat in find_repeats_n(n, lower, upper):
            result.add(repeat)
    return result

print(find_repeats('998','1012'))
```

```python
print(
    sum(
        sum(find_repeats_n(2, lower_g, upper_g))
            for lower, upper in parsed_input
            for _, lower_g, upper_g in determine_groups_n(2, lower, upper)
        )
    )
```
