```python
def parse_input(line):
    return [tuple(pair.split('-')) for pair in line.split(',')]

# parsed_input = parse_input(part1_example_input)
parsed_input = parse_input(inputdata)
```

```python
from typing import Tuple


def adjust_bounds(lower: str, upper: str) -> Tuple[str, str]:
    if len(lower) % 2:
        lower = str(10**(len(lower)))
    if len(upper) % 2:
        upper = str(10**(len(upper)-1)-1)
    return (lower, upper)

adjusted_input = [adjust_bounds(lower, upper) for lower, upper in parsed_input]
```

```python
validated_input = [(lower, upper) for lower, upper in adjusted_input if int(lower) <= int(upper)]
```

```python
for pair in validated_input:
    print(pair)
```

```python
def bisect_number(number: str) -> Tuple[str, str]:
    assert(len(number) % 2 != 1)
    half_len = len(number) // 2
    big = number[:half_len]
    small = number[half_len:]
    return (big, small)
```

```python
from typing import Iterable, List


def find_invalid(lower: str, upper: str) -> Iterable[str]:
    assert(len(lower) == len(upper))
    lower_big, lower_small = bisect_number(lower)
    upper_big, upper_small = bisect_number(upper)
    
    for i in range(
            max(int(lower_big),int(lower_small)),
            min(int(upper_big),int(upper_small))+1):
        yield str(i)*2

    return

def flatten(xxs: Iterable[Iterable[str]]) -> List[str]:
    return [x for xs in xxs for x in xs]

candidate_invalids = flatten(find_invalid(lower, upper) for lower, upper in validated_input)
for invalid in candidate_invalids:
    print(invalid)
```

```python
print(sum(int(invalid) for invalid in candidate_invalids))
```
