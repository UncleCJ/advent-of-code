```python
part1_example_input = '''123 328  51 64 
 45 64  387 23 
  6 98  215 314
*   +   *   +  '''

inputdata = downloaded['input']
```

```python
def parse_input(lines):
    number_rows = []
    for line in lines[:-1]:
        number_rows.append(list(int(number) for number in line.split(' ') if number != ''))
    operator_row = list(operator for operator in lines[-1].split(' ') if operator != '')
    results = []
    for col in range(len(number_rows[0])):
        number_col = list(number_row[col] for number_row in number_rows)
        results.append((operator_row[col], number_col))
    
    return results

parsed_expressions = parse_input(part1_example_input.splitlines())
# parsed_expressions = parse_input(inputdata.splitlines())
```

```python
from functools import reduce
import operator


result = 0
for op, numbers in parsed_expressions:
    if op == '*':
        result += reduce(operator.mul, numbers, 1)
    elif op == '+':
        result += sum(numbers)
print(result)
```

```python
def parse_input_part2(lines: list[str]) -> list[tuple[str, tuple[int, ...]]]:
    parsed_expressions = []
    
    current_operator = ' '
    current_terms = []
    for col in range(len(lines[0]) - 1, -1, -1):
        col_chars = tuple(line[col] for line in lines)
        # print(col_chars)
        if all(c == ' ' for c in col_chars):
            # Iterate
            parsed_expressions.append((current_operator, tuple(current_terms)))
            current_operator = ' '
            current_terms = []
        else:
            if col_chars[-1] != ' ':
                current_operator = col_chars[-1]
            current_terms.append(int(''.join(c for c in col_chars[:-1] if c != ' ')))
            # print(f'{current_operator=}, {current_terms[-1]}')
    if current_operator != ' ':  # _First_ expression isn't preceded by a blank column
        parsed_expressions.append((current_operator, tuple(current_terms)))

    return parsed_expressions

# parsed_expressions = parse_input_part2(part1_example_input.splitlines())
parsed_expressions = parse_input(inputdata.splitlines())
```

```python
print(parsed_expressions)
```

```python
result = 0
for op, numbers in parsed_expressions:
    if op == '*':
        result += reduce(operator.mul, numbers, 1)
    elif op == '+':
        result += sum(numbers)
print(result)
```
