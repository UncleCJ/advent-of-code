```python
part1_example_input = '''123 328  51 64 
 45 64  387 23 
  6 98  215 314
*   +   *   +  '''

inputdata = downloaded['input']
```

```python
def parse_input_chunks(lines: list[str]) -> list[list[str]]:
    def transpose(columns: list[str]) -> list[str]:
        lines = []
        for row in range(len(columns[0])):
            lines.append(''.join(column[row] for column in columns))
        return lines
        
    parsed_chunks = []
    current_columns = []
    for col in range(len(lines[0])):
        current_column = ''.join(line[col] for line in lines)
        if all(c == ' ' for c in current_column):
            parsed_chunks.append(transpose(current_columns))
            current_columns = []
        else:
            current_columns.append(current_column)
    if len(current_columns) != 0:
        parsed_chunks.append(transpose(current_columns))
    return parsed_chunks
```

```python
parsed_chunks = parse_input_chunks(part1_example_input.splitlines())
# parsed_chunks = parse_input_chunks(inputdata.splitlines())
print(parsed_chunks[:4])
```

```python
from functools import reduce
import operator


def evaluate_chunk(chunk: list[str]) -> int:
    match chunk[-1][0]:
        case '*':
            return reduce(operator.mul, (int(line) for line in chunk[:-1]))
        case '+':
            return reduce(operator.add, (int(line) for line in chunk[:-1]))
```

```python
print(sum(evaluate_chunk(chunk) for chunk in parsed_chunks))
```

```python
def evaluate_rtl_columns(chunk: list[str]) -> int:
    columns_rtl = list(''.join(line[col] for line in chunk) for col in range(len(chunk[0]) - 1, -1, -1))
    match columns_rtl[-1][-1]:
        case '+':
            return reduce(operator.add, (int(column[:-1]) for column in columns_rtl))
        case '*':
            return reduce(operator.mul, (int(column[:-1]) for column in columns_rtl))
```

```python
print(sum(evaluate_rtl_columns(chunk) for chunk in reversed(parsed_chunks)))
```
