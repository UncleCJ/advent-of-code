```python
type Junction = tuple[int, int, int]

def parse_input(lines: Iterable[str]) -> set[Junction]:
    return set(tuple(map(int, line.split(','))) for line in lines)

parsed_input = parse_input(part1_example_input.splitlines())
# parsed_input = parse_input(inputdata.splitlines())
```

```python
from itertools import combinations
from math import sqrt

type JunctionPair = tuple[Junction, Junction]

def sorted_distances(junctions: Iterable[Junction]) -> list[JunctionPair]:
    def distance(junction_a: Junction, junction_b: Junction) -> int:
        return sqrt(sum((a_coord - b_coord)**2 for a_coord, b_coord in zip(junction_a, junction_b)))
    return sorted(combinations(junctions, 2), key=lambda pair: distance(*pair))
```

```python
type Circuit = frozenset[Junction]

def connect_pair(junction_a, junction_b, circuits: set[Circuit]):
    def check_existing(junction: Junction, circuits: set[Circuit]):
        unit_circuit = frozenset((junction, ))
        if unit_circuit in circuits:
            circuits.remove(unit_circuit)
            return unit_circuit

        for circuit in circuits:
            if junction in circuit:
                circuits.remove(circuit)  # This might very well ruin further loops, but it seems to work that we break here
                return circuit
        else:
            return unit_circuit
    circuit_a = check_existing(junction_a, circuits)
    circuit_b = check_existing(junction_b, circuits)
    circuits.add(frozenset(circuit_a.union(circuit_b)))
```

```python
from functools import reduce
from operator import mul

def sort_circuits_desc(circuits: set[Circuit]) -> list[Circuit]:
    return sorted(circuits, key=lambda circuit: len(circuit), reverse=True)

circuits: set[Circuit] = set(frozenset((junction, )) for junction in parsed_input)
for _, junction_pair in zip(range(10), sorted_distances(parsed_input)):
    connect_pair(*junction_pair, circuits)

print(reduce(mul,
             [len(circuit) for circuit in sort_circuits_desc(circuits)][:min(3, len(circuits))]
             ))
```
