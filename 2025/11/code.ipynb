```python
from typing import Iterable

type DevicesOutputs = dict[str, tuple[str, ...]]

def parse_input(lines: Iterable[str]) -> DevicesOutputs:
    results: DevicesOutputs = dict()
    for line in lines:
        tokens = list(line.split(' '))
        device = tokens[0][:-1]
        outputs = tuple(tokens[1:])
        results[device] = outputs
    return results

# parsed_input = parse_input(part1_example_input.splitlines())
parsed_input = parse_input(inputdata.splitlines())
```

```python
print(parsed_input)
```

```python
from typing import Iterator


def wired_to(dos: DevicesOutputs, to_device: str) -> Iterator[str]:
    for device, outputs in dos.items():
        if to_device in outputs:
            yield device
```

```python
print(list(wired_to(parsed_input, 'out')))
```
